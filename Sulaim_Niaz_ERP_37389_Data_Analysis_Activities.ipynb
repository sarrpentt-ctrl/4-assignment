```python
# Import the libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Make charts look clear
sns.set_theme(style="whitegrid")
```

## Dataset 1 – Pakistani E-Commerce Data

File: `pakistani_ecommerce_dataset(1).csv`

Questions:
1. Most popular product category
2. Average order value
3. City with the highest number of orders
4. Most used payment method
5. Bar chart of average order value by city


```python
# Load Dataset 1
ecommerce = pd.read_csv("pakistani_ecommerce_dataset(1).csv")

# Preview the data
print(ecommerce.head())
print(ecommerce.info())
```

```python
# Q1: Most popular product category
category_counts = ecommerce["product_category"].value_counts()
print(category_counts)
print("Most popular category:", category_counts.idxmax())

# Q2: Average order value
average_order_value = ecommerce["order_value"].mean()
print("Average order value:", round(average_order_value, 2))

# Q3: City with the highest number of orders
orders_by_city = ecommerce["city"].value_counts()
print(orders_by_city)
print("City with the most orders:", orders_by_city.idxmax())

# Q4: Most used payment method
payment_counts = ecommerce["payment_method"].value_counts()
print(payment_counts)
print("Most used payment method:", payment_counts.idxmax())

# Q5: Average order value by city and bar chart
average_by_city = ecommerce.groupby("city")["order_value"].mean()
print(average_by_city.round(2))

plt.figure(figsize=(9, 5))
sns.barplot(x=average_by_city.index, y=average_by_city.values)
plt.title("Average Order Value by City")
plt.xlabel("City")
plt.ylabel("Average Order Value")
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()
```

## Dataset 2 – Student Lifestyle & Study Habits

File: `student_lifestyle_dataset.csv`

Questions:
1. Average number of study hours
2. Whether males or females sleep more on average
3. Relationship between screen time and sleep hours
4. Histogram of daily transport cost
5. Compare average internet usage by gender


```python
# Load Dataset 2
lifestyle = pd.read_csv("student_lifestyle_dataset.csv")
print(lifestyle.head())
print(lifestyle.info())
```

```python
# Q1: Average study hours
average_study_hours = lifestyle["study_hours"].mean()
print("Average study hours:", round(average_study_hours, 2))

# Q2: Average sleep hours by gender
sleep_by_gender = lifestyle.groupby("gender")["sleep_hours"].mean()
print("Average sleep hours by gender:")
print(sleep_by_gender.round(2))
print("Gender sleeping more on average:", sleep_by_gender.idxmax())

# Q3: Relationship between screen time and sleep hours
correlation = lifestyle["screen_time"].corr(lifestyle["sleep_hours"])
print("Correlation between screen time and sleep hours:", round(correlation, 3))

plt.figure(figsize=(7, 5))
sns.scatterplot(data=lifestyle, x="screen_time", y="sleep_hours")
plt.title("Screen Time vs Sleep Hours")
plt.xlabel("Screen Time")
plt.ylabel("Sleep Hours")
plt.tight_layout()
plt.show()

# Q4: Histogram of daily transport cost
plt.figure(figsize=(8, 5))
sns.histplot(data=lifestyle, x="daily_transport_cost", bins=10, kde=False)
plt.title("Distribution of Daily Transport Cost")
plt.xlabel("Daily Transport Cost")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()

# Q5: Average internet usage by gender
internet_by_gender = lifestyle.groupby("gender")["internet_usage"].mean()
print("Average internet usage by gender:")
print(internet_by_gender.round(2))

plt.figure(figsize=(7, 5))
sns.barplot(x=internet_by_gender.index, y=internet_by_gender.values)
plt.title("Average Internet Usage by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Internet Usage")
plt.tight_layout()
plt.show()
```

## Dataset 3 – Ride-Hailing in Pakistan

File: `ride_hailing_dataset.csv`

Questions:
1. Average fare by service type
2. City with the most trips
3. Average distance per trip
4. Compare trip ratings by gender
5. Bar chart of service type vs average fare


```python
# Load Dataset 3
ride = pd.read_csv("ride_hailing_dataset.csv")
print(ride.head())
print(ride.info())
```

```python
# Q1: Average fare by service type
fare_by_service = ride.groupby("service_type")["fare"].mean()
print("Average fare by service type:")
print(fare_by_service.round(2))

# Q2: City with the most trips
trips_by_city = ride["city"].value_counts()
print(trips_by_city)
print("City with the most trips:", trips_by_city.idxmax())

# Q3: Average distance per trip
average_distance = ride["distance_km"].mean()
print("Average distance per trip (km):", round(average_distance, 2))

# Q4: Compare trip ratings by gender
ratings_by_gender = ride.groupby("gender")["trip_rating"].mean()
print("Average trip rating by gender:")
print(ratings_by_gender.round(2))

# Q5: Bar chart of service type vs average fare
plt.figure(figsize=(8, 5))
sns.barplot(x=fare_by_service.index, y=fare_by_service.values)
plt.title("Average Fare by Ride-Hailing Service")
plt.xlabel("Service Type")
plt.ylabel("Average Fare")
plt.tight_layout()
plt.show()
```

## Dataset 4 – Student Tea & Snack Preferences

File: `student_tea_snacks _1_.csv`

Questions:
1. Most preferred tea type
2. Average number of cups per gender
3. Bar chart of snack choices
4. Compare tea preference by age group
5. Count students who do not drink tea

For Q1 and Q4, blank `tea_type` values are excluded because students who answered “No” to `prefers_tea` have no tea type recorded.


```python
# Load Dataset 4
tea = pd.read_csv("student_tea_snacks _1_.csv")
print(tea.head())
print(tea.info())
```

```python
# Q1: Most preferred tea type (ignore blank tea_type entries)
tea_type_counts = tea["tea_type"].dropna().value_counts()
print(tea_type_counts)
print("Most preferred tea type:", tea_type_counts.idxmax())

# Q2: Average number of cups per gender
cups_by_gender = tea.groupby("gender")["cups_per_day"].mean()
print("Average cups per day by gender:")
print(cups_by_gender.round(2))

# Q3: Bar chart of snack choices
snack_counts = tea["snack_choice"].value_counts()
print(snack_counts)

plt.figure(figsize=(7, 5))
sns.barplot(x=snack_counts.index, y=snack_counts.values)
plt.title("Student Snack Choices")
plt.xlabel("Snack Choice")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()

# Q4: Compare tea preference by age group
tea_preference_by_age = pd.crosstab(tea["age_group"], tea["tea_type"])
print("Tea type counts by age group:")
print(tea_preference_by_age)

tea_preference_by_age.plot(kind="bar", figsize=(8, 5))
plt.title("Tea Preference by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Q5: Count students who do not drink tea
non_tea_drinkers = (tea["prefers_tea"].str.strip().str.lower() == "no").sum()
print("Number of students who do not drink tea:", non_tea_drinkers)
```
